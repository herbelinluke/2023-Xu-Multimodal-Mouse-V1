```python
import glob
import pickle
import os
import re
import matplotlib.pyplot as plt
```

## Epoch vs validation correlation (score)

```python
data_dir = "epoch_vs_score_data"
key = "val_cor_list"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))
if not paths:
    print(f"No .pkl files in {data_dir}/. Run training/export first.")
else:
    fig, ax = plt.subplots(figsize=(10, 6))
    for path in paths:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key not in d:
            continue
        epochs = range(len(d[key]))
        label = "{} {} shifter={}".format(d.get("model_name", "?"), d.get("file_id", "?"), d.get("shifter", "?"))
        m = re.search(r"maxsamp_(\d+)", os.path.basename(path))
        if m:
            label = "{} maxsamp={}".format(label, m.group(1))
        if d.get("vid_type"):
            label = "{} ({})".format(label, d["vid_type"])
        ax.plot(epochs, d[key], label=label)
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean validation correlation")
    ax.legend(loc="best")
    ax.grid(True, alpha=0.3)
    plt.title("Epoch vs validation correlation (score)")
    fig.tight_layout()
    plt.show()
```

## Dataset comparison (Sensorium vs Mouse)

Load all runs from `epoch_vs_score_data` and `epoch_vs_explained_variance_data` and plot **correlation** and **explained variance** with clear dataset labels: Sensorium (from `train_cnn_shifter_table_1_sensorium`) vs Mouse (from `train_cnn_shifter_mouse_vs_sensorium`) with different `max_train_samples`. Use this to see which dataset the model trains on and predicts better.

```python
def _dataset_label(d, path):
    """Legend label inferred from meta or filename."""
    base = os.path.basename(path)
    sh = d.get("shifter", None)
    if sh is None:
        sh = True if "shifter_True" in base else False if "shifter_False" in base else False
    sh_str = "shifter=T" if sh else "shifter=F"

    ds = d.get("dataset")
    if ds is None:
        if "sensorium" in base:
            ds = "sensorium"
        elif "_mouse_" in base:
            ds = "mouse"

    if ds == "mouse":
        m = d.get("max_train_samples")
        if m is None:
            mm = re.search(r"maxsamp_(\d+)", base)
            m = int(mm.group(1)) if mm else None
        m_str = "full" if m is None else m
        return "Mouse maxsamp={} {}".format(m_str, sh_str)

    return "Sensorium {}".format(sh_str)

# Epoch vs correlation (comparison)
data_dir = "epoch_vs_score_data"
key = "val_cor_list"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))
if not paths:
    print("No .pkl files in {}. Run training first.".format(data_dir))
else:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    # Left: correlation
    ax = axes[0]
    for path in paths:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key not in d:
            continue
        epochs = range(len(d[key]))
        ax.plot(epochs, d[key], label=_dataset_label(d))
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean validation correlation")
    ax.legend(loc="best", fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_title("Epoch vs validation correlation")
    # Right: explained variance
    data_dir_ev = "epoch_vs_explained_variance_data"
    key_ev = "val_explained_var_list"
    paths_ev = sorted(glob.glob(os.path.join(data_dir_ev, "*.pkl")))
    ax = axes[1]
    for path in paths_ev:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key_ev not in d:
            continue
        epochs = range(len(d[key_ev]))
        ax.plot(epochs, d[key_ev], label=_dataset_label(d, path))
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean explained variance")
    ax.legend(loc="best", fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_title("Epoch vs explained variance")
    fig.suptitle("Dataset comparison: Sensorium vs Mouse (by max training samples)", fontsize=11)
    fig.tight_layout()
    plt.show()
```

## Per-neuron correlation histograms (by epoch)

Load per-neuron correlation from `epoch_vs_per_neuron_data/` and plot histograms by epoch. Set `epoch_to_plot` to an int to plot one epoch, or `None` to plot all epochs.

```python
import numpy as np

data_dir = "epoch_vs_per_neuron_data"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))

# Set to an int to force a single epoch; otherwise we show every `epoch_step` epochs
epoch_to_plot = None
epoch_step = 10

if not paths:
    print("No .pkl files in {}. Run training that saves per-neuron data first.".format(data_dir))
else:
    for path in paths:
        base = os.path.basename(path)
        with open(path, "rb") as f:
            d = pickle.load(f)
        cor_list = d.get("cor_per_neuron_per_epoch")
        if cor_list is None:
            continue

        # Infer dataset + maxsamp from filename (per-neuron pkls don't always store these)
        dataset = "sensorium" if "sensorium" in base else "mouse" if "_mouse_" in base else "unknown"
        mm = re.search(r"maxsamp_(\d+)", base)
        max_samp = int(mm.group(1)) if mm else None

        # Only keep Mouse CNN with maxsamp=500 and Sensorium CNN runs
        if dataset == "mouse":
            if max_samp != 500:
                continue
        elif dataset != "sensorium":
            continue

        if epoch_to_plot is None:
            epochs_to_show = list(range(0, len(cor_list), epoch_step))
        else:
            if epoch_to_plot >= len(cor_list) or cor_list[epoch_to_plot] is None:
                continue
            epochs_to_show = [epoch_to_plot]

        if not epochs_to_show:
            continue

        ncols = (len(epochs_to_show) + 1) // 2
        fig, axes = plt.subplots(2, ncols, figsize=(4 * ncols, 6))
        if len(epochs_to_show) == 1:
            axes = [axes]
        axes = np.atleast_1d(axes).flatten()

        for idx, ep in enumerate(epochs_to_show):
            if ep >= len(cor_list) or cor_list[ep] is None:
                continue
            arr = np.asarray(cor_list[ep], dtype=np.float64)
            arr = arr[~np.isnan(arr)]
            if idx < len(axes):
                axes[idx].hist(arr, bins=30, edgecolor="black", alpha=0.7)
                axes[idx].set_title("Epoch {}".format(ep))
                axes[idx].set_xlabel("Correlation")

        for j in range(len(epochs_to_show), len(axes)):
            axes[j].set_visible(False)

        sh = d.get("shifter")
        if sh is None:
            sh = True if "shifter_True" in base else False if "shifter_False" in base else "?"

        if dataset == "mouse":
            label = "Mouse CNN maxsamp=500 shifter={}".format(sh)
        else:
            label = "Sensorium CNN shifter={}".format(sh)

        fig.suptitle("Per-neuron correlation: " + label)
        fig.tight_layout()
        plt.show()
```
