## Epoch vs validation correlation (score)

```python
data_dir = "epoch_vs_score_data"
key = "val_cor_list"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))
if not paths:
    print(f"No .pkl files in {data_dir}/. Run training/export first.")
else:
    fig, ax = plt.subplots(figsize=(10, 6))
    for path in paths:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key not in d:
            continue
        epochs = range(len(d[key]))
        label = "{} {} shifter={}".format(d.get("model_name", "?"), d.get("file_id", "?"), d.get("shifter", "?"))
        m = re.search(r"maxsamp_(\d+)", os.path.basename(path))
        if m:
            label = "{} maxsamp={}".format(label, m.group(1))
        if d.get("vid_type"):
            label = "{} ({})".format(label, d["vid_type"])
        ax.plot(epochs, d[key], label=label)
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean validation correlation")
    ax.legend(loc="best")
    ax.grid(True, alpha=0.3)
    plt.title("Epoch vs validation correlation (score)")
    fig.tight_layout()
    plt.show()
```

## Dataset comparison (Sensorium vs Mouse)

Load all runs from `epoch_vs_score_data` and `epoch_vs_explained_variance_data` and plot **correlation** and **explained variance** with clear dataset labels: Sensorium (from `train_cnn_shifter_table_1_sensorium`) vs Mouse (from `train_cnn_shifter_mouse_vs_sensorium`) with different `max_train_samples`. Use this to see which dataset the model trains on and predicts better.

```python
def _dataset_label(d, path):
    """Legend label inferred from meta or filename."""
    base = os.path.basename(path)
    sh = d.get("shifter", None)
    if sh is None:
        sh = True if "shifter_True" in base else False if "shifter_False" in base else False
    sh_str = "shifter=T" if sh else "shifter=F"

    ds = d.get("dataset")
    if ds is None:
        if "sensorium" in base:
            ds = "sensorium"
        elif "_mouse_" in base:
            ds = "mouse"

    if ds == "mouse":
        m = d.get("max_train_samples")
        if m is None:
            mm = re.search(r"maxsamp_(\d+)", base)
            m = int(mm.group(1)) if mm else None
        m_str = "full" if m is None else m
        mode = d.get("training_mode")
        if mode == "transfer_frozen_backbone":
            src = d.get("pretrained_source") or "?"
            return "Mouse transfer from {} maxsamp={} {}".format(src, m_str, sh_str)
        if mode == "from_scratch" or "baseline_scratch" in base:
            return "Mouse scratch maxsamp={} {}".format(m_str, sh_str)
        if "transfer_from" in base and mode is None:
            return "Mouse transfer maxsamp={} {}".format(m_str, sh_str)
        return "Mouse maxsamp={} {}".format(m_str, sh_str)

    return "Sensorium {}".format(sh_str)

# Epoch vs correlation (comparison)
data_dir = "epoch_vs_score_data"
key = "val_cor_list"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))
if not paths:
    print("No .pkl files in {}. Run training first.".format(data_dir))
else:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    # Left: correlation
    ax = axes[0]
    for path in paths:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key not in d:
            continue
        epochs = range(len(d[key]))
        ax.plot(epochs, d[key], label=_dataset_label(d, path))
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean validation correlation")
    ax.legend(loc="best", fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_title("Epoch vs validation correlation")
    # Right: explained variance
    data_dir_ev = "epoch_vs_explained_variance_data"
    key_ev = "val_explained_var_list"
    paths_ev = sorted(glob.glob(os.path.join(data_dir_ev, "*.pkl")))
    ax = axes[1]
    for path in paths_ev:
        with open(path, "rb") as f:
            d = pickle.load(f)
        if key_ev not in d:
            continue
        epochs = range(len(d[key_ev]))
        ax.plot(epochs, d[key_ev], label=_dataset_label(d, path))
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Mean explained variance")
    ax.legend(loc="best", fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_title("Epoch vs explained variance")
    fig.suptitle("Dataset comparison: Sensorium vs Mouse (by max training samples)", fontsize=11)
    fig.tight_layout()
    plt.show()
```

## Training data volume (minutes) vs validation metrics

**Time per training sample (for converting `max_train_samples` / full-train counts):**

- **Mouse** (segmented data, `MouseDatasetSegNewBehav`): **48 ms** per frame → \(\text{minutes} = N \times 0.048 / 60\).
- **Sensorium** (`vid_frame="per_frame"`, 30 Hz stimuli / resampled traces): **1/30 s** per frame → \(\text{minutes} = N / 1800\).

**Full-train defaults when pickles omit `max_train_samples`:** mouse **30,120** (10 segments × 80% split of each segment’s train split); Sensorium **random 70/30** pool **17,917** train samples; Sensorium **tiers** train-only **12528** (348 train trials × 36 frames/trial — adjust if your release differs).

Below: scatter of **effective training minutes** vs **best validation correlation** and vs **minimum validation loss** for CNN runs in `epoch_vs_score_data/` (Sensorium + mouse, including `maxsamp_*` curves).

```python
# Training minutes vs best val correlation / min val loss (CNN runs in epoch_vs_score_data)
import numpy as np
from matplotlib.patches import Patch

MOUSE_MS_PER_SAMPLE = 48e-3
SENSORIUM_FPS = 30.0
MOUSE_FULL_TRAIN_SAMPLES = 30120
SENSORIUM_RANDOM70_TRAIN_SAMPLES = 17917
SENSORIUM_TIERS_TRAIN_SAMPLES = 348 * 36  # train tier: 348 trials × 36 frames/trial


def _infer_dataset(d, path):
    base = os.path.basename(path)
    ds = d.get("dataset")
    if ds:
        return ds
    if "sensorium" in base:
        return "sensorium"
    if "_mouse_" in base:
        return "mouse"
    # Legacy full-mouse CNN pickles: epoch_vs_score_cnn_<file_id>_vid_... without "mouse" in name
    if re.match(r"epoch_vs_score_cnn_[^s]", base) and "sensorium" not in base:
        return "mouse"
    return None


def effective_train_samples(d, path):
    """Return (dataset, n_train_samples) or (None, None) if unknown."""
    m = d.get("max_train_samples")
    ds = _infer_dataset(d, path)
    base = os.path.basename(path)
    if m is not None:
        if ds is None:
            if "_mouse_" in base:
                ds = "mouse"
            elif "sensorium" in base:
                ds = "sensorium"
        return ds, m
    if ds == "mouse":
        return "mouse", MOUSE_FULL_TRAIN_SAMPLES
    if ds == "sensorium":
        strat = d.get("split_strategy") or ""
        if strat == "random" or "random70" in base:
            return "sensorium", SENSORIUM_RANDOM70_TRAIN_SAMPLES
        if strat == "tiers" or "tiers" in base:
            return "sensorium", SENSORIUM_TIERS_TRAIN_SAMPLES
        return "sensorium", SENSORIUM_RANDOM70_TRAIN_SAMPLES
    return None, None


def train_minutes(dataset, n_samples):
    if dataset == "mouse":
        return n_samples * MOUSE_MS_PER_SAMPLE / 60.0
    if dataset == "sensorium":
        return n_samples / SENSORIUM_FPS / 60.0
    return float("nan")


def _short_label(d, path):
    base = os.path.basename(path)
    sh = d.get("shifter")
    if sh is None:
        sh = True if "shifter_True" in base else False if "shifter_False" in base else None
    sh_s = "S1" if sh else "S0"
    ds = _infer_dataset(d, path)
    if ds == "mouse":
        m = d.get("max_train_samples")
        if m is None:
            mm = re.search(r"maxsamp_(\d+)", base)
            m = int(mm.group(1)) if mm else "full"
        return "mouse {} {}".format(m, sh_s)
    if ds == "sensorium":
        st = d.get("split_strategy") or ("random70" if "random70" in base else "tiers" if "tiers" in base else "?")
        return "sensorium {} {}".format(st, sh_s)
    return base[:40]


data_dir = "epoch_vs_score_data"
paths = sorted(glob.glob(os.path.join(data_dir, "epoch_vs_score_cnn*.pkl")))
rows = []
for path in paths:
    with open(path, "rb") as f:
        d = pickle.load(f)
    if "val_cor_list" not in d or "val_loss_list" not in d:
        continue
    ds, n_samp = effective_train_samples(d, path)
    if ds is None or n_samp is None:
        continue
    mins = train_minutes(ds, n_samp)
    cor = d["val_cor_list"]
    loss = d["val_loss_list"]
    if not cor or not loss:
        continue
    best_i = int(np.argmax(cor))
    rows.append(
        {
            "path": path,
            "label": _short_label(d, path),
            "dataset": ds,
            "train_minutes": mins,
            "n_samples": n_samp,
            "best_val_cor": float(np.max(cor)),
            "min_val_loss": float(np.min(loss)),
            "val_loss_at_best_cor": float(loss[best_i]),
            "shifter": d.get("shifter"),
        }
    )

if not rows:
    print("No usable CNN pickles in {} (need val_cor_list + val_loss_list).".format(data_dir))
else:
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    ds_colors = {"mouse": "C0", "sensorium": "C1"}

    for ax, ykey, ylab in zip(
        axes,
        ("best_val_cor", "min_val_loss"),
        ("Best validation correlation (max over epochs)", "Min validation loss (min over epochs)"),
    ):
        for r in rows:
            c = ds_colors.get(r["dataset"], "gray")
            ax.scatter(
                r["train_minutes"],
                r[ykey],
                c=c,
                s=55,
                alpha=0.85,
                edgecolors="k",
                linewidths=0.4,
            )
        ax.set_xlabel("Effective training data (minutes)")
        ax.set_ylabel(ylab)
        ax.grid(True, alpha=0.3)
        leg_handles = [
            Patch(facecolor=ds_colors["mouse"], edgecolor="k", linewidth=0.4, label="mouse"),
            Patch(facecolor=ds_colors["sensorium"], edgecolor="k", linewidth=0.4, label="sensorium"),
        ]
        ax.legend(handles=leg_handles, title="dataset", fontsize=9)

    axes[0].set_title("Training minutes vs best val correlation")
    axes[1].set_title("Training minutes vs min val loss")
    fig.suptitle("CNN runs: data volume (minutes) vs validation metrics", fontsize=12)
    fig.tight_layout()
    plt.show()

    # Second figure: annotate each point (can get busy — useful for a few runs)
    fig2, ax2 = plt.subplots(figsize=(10, 6))
    for r in rows:
        c = ds_colors.get(r["dataset"], "gray")
        ax2.scatter(r["train_minutes"], r["best_val_cor"], c=c, s=50, alpha=0.8, edgecolors="k", linewidths=0.3)
        ax2.annotate(
            r["label"],
            (r["train_minutes"], r["best_val_cor"]),
            textcoords="offset points",
            xytext=(4, 4),
            fontsize=7,
            alpha=0.9,
        )
    ax2.set_xlabel("Effective training data (minutes)")
    ax2.set_ylabel("Best validation correlation")
    ax2.set_title("Annotated: training minutes vs best val correlation")
    ax2.grid(True, alpha=0.3)
    h_mouse = plt.Line2D([0], [0], color=ds_colors["mouse"], marker="o", linestyle="", label="mouse")
    h_sens = plt.Line2D([0], [0], color=ds_colors["sensorium"], marker="o", linestyle="", label="sensorium")
    ax2.legend(handles=[h_mouse, h_sens], title="dataset")
    fig2.tight_layout()
    plt.show()
```

## Per-neuron correlation histograms (by epoch)

Load per-neuron correlation from `epoch_vs_per_neuron_data/` and plot histograms by epoch. Set `epoch_to_plot` to an int to plot one epoch, or `None` to plot all epochs.

```python
import numpy as np

data_dir = "epoch_vs_per_neuron_data"
paths = sorted(glob.glob(os.path.join(data_dir, "*.pkl")))

# Set to an int to force a single epoch; otherwise we show every `epoch_step` epochs
epoch_to_plot = None
epoch_step = 10

if not paths:
    print("No .pkl files in {}. Run training that saves per-neuron data first.".format(data_dir))
else:
    for path in paths:
        base = os.path.basename(path)
        with open(path, "rb") as f:
            d = pickle.load(f)
        cor_list = d.get("cor_per_neuron_per_epoch")
        if cor_list is None:
            continue

        # Infer dataset + maxsamp from filename (per-neuron pkls don't always store these)
        dataset = "sensorium" if "sensorium" in base else "mouse" if "_mouse_" in base else "unknown"
        mm = re.search(r"maxsamp_(\d+)", base)
        max_samp = int(mm.group(1)) if mm else None

        # Only keep Mouse CNN with maxsamp=500 and Sensorium CNN runs
        if dataset == "mouse":
            if max_samp != 500:
                continue
        elif dataset != "sensorium":
            continue

        if epoch_to_plot is None:
            epochs_to_show = list(range(0, len(cor_list), epoch_step))
        else:
            if epoch_to_plot >= len(cor_list) or cor_list[epoch_to_plot] is None:
                continue
            epochs_to_show = [epoch_to_plot]

        if not epochs_to_show:
            continue

        ncols = (len(epochs_to_show) + 1) // 2
        fig, axes = plt.subplots(2, ncols, figsize=(4 * ncols, 6))
        if len(epochs_to_show) == 1:
            axes = [axes]
        axes = np.atleast_1d(axes).flatten()

        for idx, ep in enumerate(epochs_to_show):
            if ep >= len(cor_list) or cor_list[ep] is None:
                continue
            arr = np.asarray(cor_list[ep], dtype=np.float64)
            arr = arr[~np.isnan(arr)]
            if idx < len(axes):
                axes[idx].hist(arr, bins=30, edgecolor="black", alpha=0.7)
                axes[idx].set_title("Epoch {}".format(ep))
                axes[idx].set_xlabel("Correlation")

        for j in range(len(epochs_to_show), len(axes)):
            axes[j].set_visible(False)

        sh = d.get("shifter")
        if sh is None:
            sh = True if "shifter_True" in base else False if "shifter_False" in base else "?"

        if dataset == "mouse":
            label = "Mouse CNN maxsamp=500 shifter={}".format(sh)
        else:
            label = "Sensorium CNN shifter={}".format(sh)

        fig.suptitle("Per-neuron correlation: " + label)
        fig.tight_layout()
        plt.show()
```
